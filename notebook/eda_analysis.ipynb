# Factors That Influence Career Success (SDG 4 & 8)

## A. Introduction

The shift from college to the job market has become more competitive and
varied in recent times. Employers have begun to look beyond just academic success
and are now focusing on real-world experience, interpersonal skills, the ability to
build connections, and relevant certifications. This change from prioritizing degrees to
valuing skills highlights the increasing need for graduates who can swiftly adjust to
actual job challenges and make meaningful contributions to their workplaces.
Simultaneously, younger individuals entering the job market are prioritizing not only
salary but also fulfillment in their careers, opportunities for growth, and the balance
between work and personal life. Consequently, achieving success early in a career is
now shaped by a mix of academic achievements, social skills, work exposure, and
personal health. Nonetheless, even with similar educational qualifications, graduates
might face varying professional results such as differences in salaries, the quantity of
job offers, speed of promotions, and potential for career growth.


This collection of data includes 400 anonymous student records that reflect a
range of academic, professional, and personal growth metrics, such as GPA,
internship experiences, certifications, networking competence, soft skills, starting
salaries, job offers, job satisfaction, and current employment status. This information
allows for an analysis of how educational success and hands-on experience affect
early job outcomes for graduates. Furthermore, this dataset allows for the examination
of possible differences in job results among various demographic and educational
backgrounds, with a particular focus on gender equality in career prospects. By
investigating the connections between academic success, hands-on experiences,
networking skills, and career results, this research seeks to uncover which factors
have the strongest impact on job readiness, salary progression, promotion chances,
and general career satisfaction in the early phases of professional life.

## B. Data Understanding

The Education and Career Success dataset explores how students’ academic
achievements, skills, and extracurricular activities translate into real-world career
outcomes. By combining educational metrics  such as GPA, test scores, field of study
with measures of practical experience such as internships, projects, certifications and
soft-skill indicators such as networking, communication, this dataset provides a
holistic view of the factors that drive early-career success.

- Student Demographics
  - Student_ID : A unique alphanumeric code assigned to each student to ensure privacy while enabling record linkage.
  - Age : The student’s age, in whole years, at the time of graduation or data collection (range 18–30).
  - Gender :Self-reported gender identity: “Male”, “Female”, or “Other”

- Academic Performance
  - High_School_GPA : Cumulative grade-point average from high school on a 2.0–4.0 scale.
  - SAT_Score : Total score on the SAT college-entrance exam, ranging from 900 to 1600.
  -  University_GPA : Cumulative undergraduate GPA on a 2.0–4.0 scale.
  - Field_of_Study : Major or academic discipline (e.g., “Computer Science”, “Medicine”, “Business”).

- Skills & Extracurriculars
  - Internships_Completed : Number of formal internship experiences completed during college (0–4)
  - Projects_Completed : Count of significant academic or personal projects (e.g., capstones, research, apps) finished (0–9).
  - Certifications : Number of industry-recognized certificates earned (e.g., AWS, PMP) beyond formal coursework (0–5).
  - Soft_Skills_Score : Instructor-or peer-rated measure communication, teamwork, and leadership skills, on a 1–10 scale
  - Networking_Score : Quantitative assessment of professional networking activity (e.g., LinkedIn connections, attended events) on a 1–10 scale.

- Career Outcomes
  - Job_Offers : Total number of distinct full-time job offers received within six months of graduation (0–5).
  - Starting_Salary : Annual base salary in USD for the first full-time position (range \$25,000–\$150,000).
  - Career_Satisfaction : Self-reported satisfaction with first job on a 1 (very dissatisfied) to 10 (very satisfied) scale.  
  - Years_to_Promotion : Number of years elapsed from hire date to promotion to the next career level (1–5).
  - Current_Job_Level : Categorical career stage at data collection:“Entry”, “Mid”, “Senior”, or “Executive”.
  - Work_Life_Balance : Self-assessed balance between professional and personal life, rated 1 (poor) to 10 (excellent).
  -  Entrepreneurship : Indicates whether the individual has founded or co-founded a business (“Yes” or “No”).

#### b. How much do soft skills and certifications impact career success compared to education?

```python
education_cols = ["High_School_GPA", "SAT_Score", "University_GPA"]
soft_skills_cols = ["Soft_Skills_Score", "Certifications"]
career_cols = ["Starting_Salary", "Job_Offers", "Career_Satisfaction"]

df["Education"] = df[education_cols].mean(axis=1)
df["Soft Skills + Certifications"] = df[soft_skills_cols].mean(axis=1)
df["Career_Success_Score"] = df[career_cols].mean(axis=1)


selected_cols = ["Education", "Soft Skills + Certifications", "Career_Success_Score"]
df_num = df[selected_cols]

corr_spearman = df_num.corr("spearman")

plt.figure(figsize=(8, 5))
sns.heatmap(corr_spearman, annot=True, cmap="coolwarm", fmt=".4f")
plt.title("Correlation Heatmap (Aggregated Columns)")
plt.tight_layout()
plt.show()
```

```python
target = "Career_Success_Score"
target_corr = corr_spearman[target].drop(target)
target_corr = target_corr.sort_values(key=abs, ascending=False)

print("\n=== CORRELATION WITH TARGET ===")
print(target_corr)


plt.figure(figsize=(7, 5))
bars = plt.bar(
    target_corr.index, target_corr.values, color=["orange", "skyblue"]
)

for bar in bars:
    height = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        height + 0.002,
        f"{height:.4f}",
        ha="center",
    )

min_val = target_corr.min()
plt.ylim(min_val - 0.01, 1.0)

plt.title("Average Impact on Career Success")
plt.ylabel("Average Absolute Correlation")
plt.xlabel("Categories")
plt.tight_layout()
plt.show()
```

The data analysis indicates that although both groups have an extremely strong and nearly flawless positive influence on career results, Soft Skills + Certifications have a slightly higher impact than traditional Education in terms of job offers. The analysis shows that Soft Skills + Certifications demonstrate a higher average absolute correlation to career achievement at 0. 9758, compared to formal Education, which sits at 0. 9703. This small yet significant difference (0. 0055) supports the ongoing shift towards skills-based hiring on a global scale, highlighting how today’s recruiters value a candidate's versatile and practical experience more than just the reputation of their educational institution.



The heatmap matrix reveals a strong correlation (r = 0. 9787) between a person’s educational qualifications and their skillset in soft skills and certifications. This suggests that these two aspects are interconnected; candidates with solid academic achievements are very likely to enhance their education with relevant industry certifications and strong communication abilities. In summary, while having a formal degree is a crucial initial step for entering the job market, it is the excellence in soft skills and specific professional certifications that truly sets candidates apart when it comes to landing high-quality job opportunities.


### 4. Gender Parity in Professional Outcomes

#### a. Are there significant differences in starting salaries and promotion timelines between men and women in the same field of study ?

```python
target_fields = ['Engineering', 'Business']
df_filtered = df[df['Field_of_Study'].isin(target_fields)].copy()

sns.set(style="whitegrid")

plt.figure(figsize=(10, 6))
ax1 = sns.barplot(x='Field_of_Study', y='Starting_Salary', hue='Gender',
                  data=df_filtered, palette='Blues_d', ci=None)

plt.title('Average Starting Salary by Gender Across Engineering and Business Fields', fontsize=14)
plt.ylabel('Average Starting Salary ($)')
plt.xlabel('Field of Study')

for p in ax1.patches:
    ax1.annotate(f'${p.get_height():,.0f}',
                 (p.get_x() + p.get_width() / 2., p.get_height()),
                 ha = 'center', va = 'center',
                 xytext = (0, 9),
                 textcoords = 'offset points', fontsize=10)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
ax2 = sns.barplot(x='Field_of_Study', y='Years_to_Promotion', hue='Gender',
                  data=df_filtered, palette='Oranges_d', ci=None)

plt.title('Average Time to Promotion by Gender Across Engineering and Business Fields', fontsize=14)
plt.ylabel('Average Time to Promotion (Years)')
plt.xlabel('Field of Study')

for p in ax2.patches:
    ax2.annotate(f'{p.get_height():.1f} Thn',
                 (p.get_x() + p.get_width() / 2., p.get_height()),
                 ha = 'center', va = 'center',
                 xytext = (0, 9),
                 textcoords = 'offset points', fontsize=10)

plt.tight_layout()
plt.show()

summary = df_filtered.groupby(['Field_of_Study', 'Gender'])[['Starting_Salary', 'Years_to_Promotion']].mean()
print("Summary od Data Comparison:")
print(summary)
```

Across both Business and Engineering, women hold a clear advantage
in both starting salaries and promotion speeds. Women outearn men in both
fields, with the most dramatic gap in Business, where females average \$91,756
compared to males at \$64,370. This professional edge extends directly into
career progression; women earn promotions faster than men in both sectors,
most notably in Business, where women take 2.5 years on average compared
to men's 4.3 years. Ultimately, women in these fields are entering the
workforce with higher pay and climbing the corporate ladder at a quicker
pace.

#### b. To what extent does gender influence the number of job offers received by graduates, even when they have equivalent GPAs?

```python
bins = [0, 3.3, 3.6, 4.0]
labels = ['GPA 3.0 - 3.3', 'GPA 3.4 - 3.6', 'GPA 3.7 - 4.0']
df['GPA_Group'] = pd.cut(df['University_GPA'], bins=bins, labels=labels)

sns.set(style="whitegrid")
plt.figure(figsize=(10, 6))

ax = sns.barplot(x='GPA_Group', y='Job_Offers', hue='Gender', data=df, palette='viridis', ci=None)

plt.title('Average Number of Job Offers by Gender\nWithin Comparable GPA Groups', fontsize=14)
plt.ylabel('Average Number of Job Offers')
plt.xlabel('Academic Performance Group (GPA)')
plt.legend(title='Gender')

for p in ax.patches:
    ax.annotate(f'{p.get_height():.2f}',
                 (p.get_x() + p.get_width() / 2., p.get_height()),
                 ha = 'center', va = 'center',
                 xytext = (0, 9),
                 textcoords = 'offset points', fontsize=10)

plt.tight_layout()
plt.show()
```

Based on the chart, higher GPAs clearly lead to more job offers for
everyone, but the success rate between genders varies by academic tier. In the
lower and mid-range brackets (GPAs 3.0 to 3.6), women consistently
outperform men, securing a noticeably higher average of offers. However, this
gap virtually disappears in the top tier (GPAs 3.7 to 4.0), where men
marginally edge out women with 4.53 offers compared to 4.49. Overall,
women seem to have a competitive advantage earlier on, but the playing field
completely levels out among the highest achievers.

### E. Conclusion

This research provides strong empirical evidence that early-career success is
no longer determined solely by academic achievement, but rather by a critical
combination of academic performance, practical experience, soft skills, and external
factors such as gender and field of study. Although maintaining a high GPA in
university is essential for entering the job market, the nearly perfect positive
correlations found in completed projects (r = 0. 9666), internships (r = 0. 9663), and
soft skills (r = 0. 9758) highlight that today’s employers place a greater emphasis on
practical, real-world experiences over just academic qualifications. This is further
exemplified by the $18,000 salary difference between graduates with the same 3. 7
GPA. Additionally, external structural factors, such as the chosen field of study, create
significant inequality in early-career rewards, with high-paying technical majors like
Computer Science structurally offering more opportunities to build these lucrative
portfolios. This situation is made more challenging by a significant economic
trade-off (r = -0. 852), where prioritizing personal time over immediate financial
rewards severely lowers long-term job satisfaction, resulting in those who focus on
salary reporting much greater career fulfillment (9. 58 out of 10) compared to those
who emphasize a work-life balance (6. 70 out of 10). On a positive note, a promising
shift is occurring in terms of gender equity within competitive fields such as Business
and Engineering. Female graduates are breaking down long-standing biases by
obtaining more job offers, achieving higher starting salaries, and advancing in their
careers more quickly than their male peers. In conclusion, to successfully navigate the
evolving contemporary job market, higher education institutions must adapt to
project-focused learning and certification options, while graduates need to
